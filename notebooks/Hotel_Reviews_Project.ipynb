```python
hotel_df['reviews.date'] = pd.to_datetime(
    hotel_df['reviews.date'],
    errors='coerce'
)

print(hotel_df['reviews.date'].dtype)
```

```python
review_counts = (
    hotel_df
    .groupby('name')
    .size()
    .reset_index(name='review_count')
)

review_counts.head()
```

```python
average_ratings = (
    hotel_df
    .groupby('name')['reviews.rating']
    .mean()
    .reset_index(name='avg_rating')
)

average_ratings.head()
```

```python
hotel_info = (
    hotel_df[
        ['name', 'city', 'province', 'latitude', 'longitude',
         'categories', 'primaryCategories']
    ]
    .drop_duplicates('name')
)

print("Unique hotels:", len(hotel_info)) 
```

```python
hotel_data = hotel_info.merge(
    review_counts,
    on='name',
    how='left'
)

hotel_data = hotel_data.merge(
    average_ratings,
    on='name',
    how='left'
)

print("Hotel-level dataset shape:", hotel_data.shape)
```
